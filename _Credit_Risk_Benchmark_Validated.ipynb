```
from irb_scorecard.io import load_csv
df = load_csv('../data/Bank_Loan_Approval_copy.csv')
df.head()
```

```
from irb_scorecard.pd_proxy import train_pd_proxy, score_pd_proxy
pd_art = train_pd_proxy(df)
pd_art.metrics
```

```
scored_pd = score_pd_proxy(df, pd_art)
scored_pd[['prob_def','loan_approval','EAD','LGD','EL']].head()
```

```
from irb_scorecard.scorecard_woe import train_scorecard, predict_pbad_and_score, build_points_table
sc_art = train_scorecard(df)
scored_sc, Xw = predict_pbad_and_score(df, sc_art)
build_points_table(sc_art).head()
```

```
from irb_scorecard.policy import add_bands, add_pricing
from irb_scorecard.reason_codes import add_reason_codes
from irb_scorecard.portfolio import add_ead_lgd_el, portfolio_summary
scored_sc = add_bands(scored_sc)
scored_sc = add_pricing(scored_sc, pd_col='prob_def_proxy')
scored_sc = add_reason_codes(scored_sc, Xw, sc_art)
scored_sc = add_ead_lgd_el(scored_sc)
portfolio_summary(scored_sc)
```
